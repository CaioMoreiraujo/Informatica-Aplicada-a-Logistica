```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# URL do arquivo CSV bruto (raw) no GitHub
url = 'https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv'

# Carregar os dados
df = pd.read_csv(url)

# Exibir as primeiras linhas e colunas para verificar os nomes exatos
display(df.head())
print("Colunas disponíveis:", df.columns.tolist())
```

```python
# Criar o gráfico de dispersão (Scatter Plot) com os nomes de colunas corretos (em minúsculas)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score', alpha=0.7, color='b')

plt.title('Relação entre Horas Estudadas e Notas do Exame')
plt.xlabel('Horas Estudadas (X)')
plt.ylabel('Nota do Exame (Y)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
import numpy as np
from scipy import stats

# Calcular a regressão linear
x = df['hours_studied']
y = df['exam_score']
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

# Coeficiente de determinação (R²)
r_squared = r_value ** 2

# Criar o gráfico
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score', alpha=0.7, color='b', label='Dados reais')

# Adicionar a linha de tendência
val_x = np.linspace(x.min(), x.max(), 100)
val_y = slope * val_x + intercept
plt.plot(val_x, val_y, color='red', linewidth=2, label='Linha de Tendência')

plt.title('Relação entre Horas Estudadas e Notas do Exame (com Regressão)')
plt.xlabel('Horas Estudadas (X)')
plt.ylabel('Nota do Exame (Y)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# Exibir a equação e o valor do R² abaixo do gráfico
print(f"Equação da reta: Y = {slope:.4f} * X + ({intercept:.4f})")
print(f"Coeficiente de determinação (R²): {r_squared:.4f}")
```
